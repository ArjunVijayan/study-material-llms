```python
import transformers
from transformers import DistilBertTokenizer
```

```python
tokenizer = DistilBertTokenizer.from_pretrained('distilbert-base-uncased')


input = tokenizer.encode(
    "Hello, how are you?",
    add_special_tokens=True,
    max_length=20,
    padding='max_length',
    truncation=True,
    return_token_type_ids=True,
    return_attention_mask=True,
)

print("Input IDs:", input)
```

```python
inputs = tokenizer(
    "Hello, how are you. I love programming in Python. It is a great language for data science and machine learning.",
    add_special_tokens=True,
    max_length=50,
    padding="max_length",
    truncation=True,
    return_attention_mask=True,
    return_token_type_ids=True)

print('input_ids:', inputs['input_ids'])
print('attention_mask:', inputs['attention_mask'])
print('token_type_ids:', inputs['token_type_ids'])
```
